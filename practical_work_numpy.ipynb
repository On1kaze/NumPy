## Задание 2. Специальные массивы
**Условие:** Создайте:
а) массив из пяти нулей;
б) массив из четырёх единиц;
в) последовательность чётных чисел от 2 до 12 включительно.

````python
# а) массив из пяти нулей
zeros_arr = np.zeros(5)
print("а) Массив из пяти нулей:", zeros_arr)

# б) массив из четырёх единиц
ones_arr = np.ones(4)
print("б) Массив из четырёх единиц:", ones_arr)

# в) последовательность чётных чисел от 2 до 12 включительно
# np.arange(start, stop, step): 13 указывается, чтобы 12 вошло в результат
even_seq = np.arange(2, 13, 2)
print("в) Чётные числа от 2 до 12 включительно:", even_seq)
````

## Задание 3. Свойства массива
**Условие:** Для массива ниже определите без запуска программы значения `ndim`, `shape`, `size` и `dtype`. Затем проверьте ответ в Python.
```python
a = np.array([[3, 6, 9],
              [12, 15, 18]])
```

### Теоретический расчёт (без запуска программы):
- **`ndim`** (количество измерений/осей): `2` (двумерный массив / матрица)
- **`shape`** (форма / кортеж размеров): `(2, 3)` (2 строки, 3 столбца)
- **`size`** (общее количество элементов): `6` ($2 \times 3 = 6$)
- **`dtype`** (тип данных): `int32` или `int64` (целые числа по умолчанию)

````python
a = np.array([[3, 6, 9],
              [12, 15, 18]])

print("Проверка в Python:")
print("• a.ndim:", a.ndim)
print("• a.shape:", a.shape)
print("• a.size:", a.size)
print("• a.dtype:", a.dtype)
````

## Задание 4. Индексы
**Условие:** Дан массив `numbers`. Выведите первый, третий и последний элементы. После этого замените число 30 на 100.
```python
numbers = np.array([10, 20, 30, 40, 50])
```

````python
numbers = np.array([10, 20, 30, 40, 50])
print("Исходный массив:", numbers)

# Индексация начинается с 0
print("Первый элемент (индекс 0):", numbers[0])
print("Третий элемент (индекс 2):", numbers[2])
print("Последний элемент (индекс -1):", numbers[-1])

# Замена числа 30 на 100 (число 30 имеет индекс 2)
numbers[2] = 100
print("Массив после замены 30 на 100:", numbers)
````

## Задание 5. Срезы
**Условие:** Из массива `numbers` получите:
а) элементы с индексами от 1 до 3;
б) первые три элемента;
в) элементы начиная с индекса 2.

````python
numbers = np.array([10, 20, 30, 40, 50])
print("Исходный массив:", numbers)

# а) элементы с индексами от 1 до 3 (индексы 1, 2, 3 включительно: [1:4])
print("а) Элементы с индексами от 1 до 3 [1:4]:", numbers[1:4])
print("   (вариант срез 1:3):", numbers[1:3])

# б) первые три элемента (индексы 0, 1, 2)
print("б) Первые три элемента [:3]:", numbers[:3])

# в) элементы начиная с индекса 2 (до конца массива)
print("в) Элементы начиная с индекса 2 [2:]:", numbers[2:])
````

## Задание 6. Двумерный массив
**Условие:** Для матрицы `grades` выведите оценку 88, всю вторую строку и весь третий столбец.
```python
grades = np.array([[85, 90, 78],
                   [70, 88, 92],
                   [95, 81, 89]])
```

````python
grades = np.array([[85, 90, 78],
                   [70, 88, 92],
                   [95, 81, 89]])
print("Матрица grades:")
print(grades)

# Оценка 88 находится во 2-й строке (индекс 1) и 2-м столбце (индекс 1)
print("• Оценка 88:", grades[1, 1])

# Вся вторая строка (индекс 1)
print("• Вся вторая строка:", grades[1])

# Весь третий столбец (индекс 2)
print("• Весь третий столбец:", grades[:, 2])
````
